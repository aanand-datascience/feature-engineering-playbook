```python
### Independent and Dependent Features
x=df.drop('Class',axis=1)
y=df['Class']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.7,random_state=42)
```

**Cross Validation Like KFOLD and Hyperparameter Tuning**

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.model_selection import KFold
from sklearn.model_selection import GridSearchCV
import numpy as np
```

```python
10.0** np.arange(-2,3)
```

```python
log_class=LogisticRegression()
grid={'C':10.0 **np.arange(-2,3),'penalty':['l2']}
cv=KFold(n_splits=5,random_state=None,shuffle=False)
```

```python
# import pandas as pd

# Combine x_train and y_train to ensure row-wise NaN handling
# train_df = pd.concat([x_train, y_train], axis=1)

# Drop rows with any NaN values
# train_df_cleaned = train_df.dropna()

# Split back into x_train and y_train
# x_train_cleaned = train_df_cleaned.drop('Class', axis=1)
# y_train_cleaned = train_df_cleaned['Class']

# Now fit GridSearchCV with the cleaned data
clf = GridSearchCV(log_class, grid, cv=cv, n_jobs=-1, scoring='f1_macro')
clf.fit(x_train, y_train)
```

```python
y_pred=clf.predict(x_test)
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

##**Class_weight**
**Use this techinique (Class_weight) when your dataset is to much long(lot of imbalanced values)**

```python
# Use when you have lot of imbalanced data
class_weight=dict({0:1,1:100})
```

```python
from sklearn.ensemble import RandomForestClassifier
classifier=RandomForestClassifier(class_weight=class_weight)
classifier.fit(x_train,y_train)
```

```python
y_pred=classifier.predict(x_test)
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

## **Under Sampling**

**Never do Under Sampling unless or untill your dataset is vary-vary less**

```python
y_train.value_counts()
```

```python
from collections import Counter
from imblearn.under_sampling import NearMiss
ns=NearMiss(sampling_strategy=0.8)
x_train_ns,y_train_ns=ns.fit_resample(x_train,y_train)
print('The number of classes before fit: {}'.format(Counter(y_train)))
print('The number of classes after fit: {}'.format(Counter(y_train_ns)))
```

```python
from sklearn.ensemble import RandomForestClassifier
classifier=RandomForestClassifier()
classifier.fit(x_train_ns,y_train_ns)
```

```python
y_pred=classifier.predict(x_test)
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

##**Over Sampling**

```python
from imblearn.over_sampling import RandomOverSampler
```

```python
os=RandomOverSampler(sampling_strategy=0.75)
x_train_os,y_train_os=os.fit_resample(x_train,y_train)
print("The number of classes before fit{}".format(Counter(y_train)))
print('The number of classes after fit{} '.format(Counter(y_train_os)))
```

```python
from sklearn.ensemble import RandomForestClassifier
classifier=RandomForestClassifier()
classifier.fit(x_train_os,y_train_os)
```

```python
y_pred=classifier.predict(x_test)
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

**Important Interview Question:-> How to Handle Imblanced data-set**

"If I get an imbalanced dataset, I would first understand the class distribution and the business problem. I wouldn't rely only on accuracy; I would evaluate precision, recall, F1-score, ROC-AUC and, where appropriate, PR-AUC.

Then, based on the dataset size and imbalance ratio, I could try techniques such as under-sampling, over-sampling or SMOTE. I would be careful with under-sampling because it removes observations.

I would also use class-weighted models such as Random Forest or XGBoost. Based on domain knowledge, I would determine whether false positives or false negatives are more costly. Then I could tune the classification threshold—for example, from 0.5 to 0.6 or 0.7—to achieve the desired precision-recall trade-off. Finally, I would compare these approaches using cross-validation and select the approach that gives the best business-relevant performance."
